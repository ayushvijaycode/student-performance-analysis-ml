# Student Performance Analysis & Score Prediction

A small end-to-end Data Analysis + Machine Learning project. We explore student academic/lifestyle data and predict final exam scores using Linear Regression.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../data/student_performance.csv')
df.head()
```

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
print('\nSummary statistics:')
df.describe().round(2)
```

## 1. Data Cleaning

Missing numeric values are replaced with their median values.

```python
features = ['study_hours_per_day', 'attendance_percent', 'sleep_hours',
            'assignment_completion_percent', 'previous_exam_score']
target = 'final_exam_score'

df[features + [target]] = df[features + [target]].apply(lambda col: col.fillna(col.median()))
df.isnull().sum()
```

## 2. Exploratory Data Analysis

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='study_hours_per_day', y='final_exam_score')
plt.title('Study Hours vs Final Exam Score')
plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.heatmap(df[features + [target]].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## 3. Build the Machine Learning Model

```python
X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

print(f'MAE  : {mae:.2f}')
print(f'RMSE : {rmse:.2f}')
print(f'R²   : {r2:.3f}')
```

```python
coefficients = pd.Series(model.coef_, index=features).sort_values(key=abs, ascending=False)
coefficients
```

```python
plt.figure(figsize=(7,6))
plt.scatter(y_test, y_pred, alpha=0.75)
lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
plt.plot(lims, lims, '--')
plt.xlabel('Actual Score')
plt.ylabel('Predicted Score')
plt.title('Actual vs Predicted Final Exam Score')
plt.show()
```

## 4. Conclusion

The model uses study time, attendance, sleep, assignment completion and previous exam performance to estimate the final exam score. The notebook demonstrates a complete beginner-friendly ML workflow: loading data → cleaning → EDA → feature selection → training → evaluation → interpretation.
